```python
%sql
CREATE OR REPLACE TABLE workspace.silver.dicionario
USING DELTA AS
SELECT DISTINCT
    CAST(id_tabela AS STRING) AS id_tabela,
    TRIM(nome_coluna) AS nome_coluna,
    CAST(chave AS STRING) AS chave,
    TRIM(valor) AS valor,
    cobertura_temporal
FROM workspace.bronze.dicionario_raw;
```
