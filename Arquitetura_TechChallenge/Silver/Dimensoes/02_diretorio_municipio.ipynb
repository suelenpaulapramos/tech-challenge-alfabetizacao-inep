# Silver | Tratamento: Diretório de Municípios
Seleciona id_municipio e nome do município do diretório oficial da Base dos Dados.
Origem: `workspace.bronze.diretorio_municipio_raw` (ingestão 08).
Chave de negócio: `id_municipio`.

```python
%run ../../Monitoring/_common_monitoring
```

```python
_m = iniciar_etapa()
```

```python
%sql

CREATE OR REPLACE TABLE workspace.silver.diretorio_municipio
USING DELTA AS
SELECT
    CAST(id_municipio AS STRING) AS id_municipio,
    TRIM(nome) AS nome_municipio,
    data_ingestao,
    fonte,
    lote_id
FROM workspace.bronze.diretorio_municipio_raw
WHERE id_municipio IS NOT NULL
QUALIFY ROW_NUMBER() OVER (
    PARTITION BY id_municipio
    ORDER BY data_ingestao DESC
) = 1;
```

```python
registrar_execucao("SILVER", "diretorio_municipio", _m, row_count=contar_tabela("workspace.silver.diretorio_municipio"))
```
